```python
print("Columns:", dataset.columns.tolist())
display(dataset.head())

text_candidates = ['text','Text','sentence','Sentence','content','tweet','Tweet','message','Message','cleaned_text']
label_candidates = ['label','Label','emotion','Emotion','sentiment','Sentiment','target']

text_col = None
label_col = None

for c in text_candidates:
    if c in dataset.columns:
        text_col = c
        break
for c in label_candidates:
    if c in dataset.columns:
        label_col = c
        break

# fallback: assume first column is text and second is label
if text_col is None:
    text_col = dataset.columns[0]
if label_col is None:
    label_col = dataset.columns[1] if len(dataset.columns) > 1 else dataset.columns[0]

print("Using text column:", text_col)
print("Using label column:", label_col)
```

```python
label_counts = dataset["label"].value_counts()
print(label_counts)
```

```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

nltk.download('stopwords')
stop_words = set(stopwords.words('english'))
stemmer = PorterStemmer()
```

```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r'https?://\S+|www\.\S+', '', text) # Remove URLs
    text = re.sub(r'<.*?>', '', text) # Remove HTML tags
    text = re.sub(r'[^\w\s]', '', text) # Remove punctuation
    text = re.sub(r'\d+', '', text) # Remove numbers
    text = ' '.join([word for word in text.split() if word not in stop_words])
    text = ' '.join([stemmer.stem(word) for word in text.split()])
    return text
```

```python
dataset['cleaned_text'] = dataset['text'].apply(clean_text)
```

```python
from sklearn.preprocessing import LabelEncoder
label_encoder = LabelEncoder()
dataset['encoded_label'] = label_encoder.fit_transform(dataset[label_col])
print("\nOriginal labels to integer mapping:")
for i, label in enumerate(label_encoder.classes_):
    print(f"{label}: {i}")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer(max_features=5000)
X = tfidf_vectorizer.fit_transform(dataset['cleaned_text'])
y = dataset['encoded_label']
```

```python
from imblearn.over_sampling import SMOTE
from collections import Counter
print("Original dataset shape:", Counter(y))
smote = SMOTE(random_state=42)
X_res, y_res = smote.fit_resample(X, y)
print("Resampled dataset shape:", Counter(y_res))
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_res, y_res, test_size=0.2, random_state=42, stratify=y_res)
```

```python
from sklearn.naive_bayes import MultinomialNB
nb_model = MultinomialNB()
nb_model.fit(X_train, y_train)
```

```python
from sklearn.metrics import classification_report, accuracy_score
y_pred = nb_model.predict(X_test)
new_accuracy = accuracy_score(y_test, y_pred)
print(f"New Model Accuracy: {new_accuracy:.4f}")
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```
